```python
import pandas as pd

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity


# ==========================================================
# SISTEM REKOMENDASI MATA KULIAH
# CONTENT-BASED FILTERING + COSINE SIMILARITY
# ==========================================================


# ==========================================================
# DATASET
# ==========================================================

data = {

    'mata_kuliah': [

        'Pemrograman Web',
        'Basis Data',
        'Pemrograman Mobile',
        'Rekayasa Perangkat Lunak',
        'Jaringan Komputer',
        'Sistem Informasi',
        'Data Mining',
        'Kecerdasan Buatan',
        'Analisis dan Perancangan Sistem',
        'Keamanan Informasi'

    ],

    'deskripsi': [

        # 1. Pemrograman Web
        'pemrograman web HTML CSS JavaScript frontend backend website',

        # 2. Basis Data
        'database SQL MySQL tabel query data relasional penyimpanan data',

        # 3. Pemrograman Mobile
        'pemrograman mobile Android Java Kotlin aplikasi smartphone',

        # 4. Rekayasa Perangkat Lunak
        'rekayasa perangkat lunak pengembangan software testing debugging pemrograman',

        # 5. Jaringan Komputer
        'jaringan komputer TCP IP router switch internet protokol komunikasi jaringan',

        # 6. Sistem Informasi
        'sistem informasi bisnis organisasi manajemen informasi database data',

        # 7. Data Mining
        'data mining data analisis klasifikasi clustering database pola informasi',

        # 8. Kecerdasan Buatan
        'kecerdasan buatan artificial intelligence machine learning algoritma klasifikasi',

        # 9. Analisis dan Perancangan Sistem
        'analisis perancangan sistem kebutuhan pengguna proses bisnis pemodelan sistem',

        # 10. Keamanan Informasi
        'keamanan informasi cyber security jaringan enkripsi firewall serangan data'

    ]
}


# ==========================================================
# KETERANGAN DATASET
# ==========================================================

keterangan_dataset = {

    'Pemrograman Web':
        'Mempelajari pembuatan website menggunakan HTML, CSS, JavaScript, frontend dan backend.',

    'Basis Data':
        'Mempelajari database, SQL, MySQL, tabel, query, relasi dan penyimpanan data.',

    'Pemrograman Mobile':
        'Mempelajari pembuatan aplikasi mobile menggunakan Android, Java dan Kotlin.',

    'Rekayasa Perangkat Lunak':
        'Mempelajari proses pengembangan software, pemrograman, testing dan debugging.',

    'Jaringan Komputer':
        'Mempelajari jaringan komputer, TCP/IP, router, switch, internet dan komunikasi jaringan.',

    'Sistem Informasi':
        'Mempelajari penerapan teknologi informasi untuk kebutuhan bisnis, organisasi dan pengelolaan data.',

    'Data Mining':
        'Mempelajari pengolahan dan analisis data menggunakan klasifikasi, clustering dan pencarian pola.',

    'Kecerdasan Buatan':
        'Mempelajari artificial intelligence, machine learning dan algoritma untuk membuat sistem cerdas.',

    'Analisis dan Perancangan Sistem':
        'Mempelajari analisis kebutuhan, proses bisnis, perancangan sistem dan pemodelan sistem.',

    'Keamanan Informasi':
        'Mempelajari keamanan data dan jaringan, cyber security, enkripsi, firewall dan perlindungan informasi.'
}


# ==========================================================
# MEMBUAT DATAFRAME
# ==========================================================

df = pd.DataFrame(data)


# ==========================================================
# MENAMPILKAN DATASET
# ==========================================================

print("================================================")
print("       SISTEM REKOMENDASI MATA KULIAH")
print("================================================")

print("\nDATASET MATA KULIAH:")
print(df.to_string(index=False))


# ==========================================================
# MENAMPILKAN KETERANGAN DATASET
# ==========================================================

print("\n================================================")
print("           KETERANGAN DATASET")
print("================================================")

for mata_kuliah, keterangan in keterangan_dataset.items():

    print(f"\n{mata_kuliah}:")
    print(keterangan)


# ==========================================================
# TF-IDF
# ==========================================================

vectorizer = TfidfVectorizer()

tfidf_matrix = vectorizer.fit_transform(
    df['deskripsi']
)


print("\n================================================")
print("                PROSES TF-IDF")
print("================================================")

print("Ukuran matriks TF-IDF:")
print(tfidf_matrix.shape)


# ==========================================================
# COSINE SIMILARITY
# ==========================================================

similarity = cosine_similarity(
    tfidf_matrix
)


print("\nCosine Similarity berhasil dihitung.")


# ==========================================================
# FUNGSI REKOMENDASI
# ==========================================================

def rekomendasi_mata_kuliah(nama_mk, jumlah=5):

    # Membersihkan input
    nama_mk = nama_mk.strip().lower()

    # Membuat nama mata kuliah menjadi huruf kecil
    daftar_mk = (
        df['mata_kuliah']
        .astype(str)
        .str.strip()
        .str.lower()
    )

    # Mengecek apakah mata kuliah tersedia
    if nama_mk not in daftar_mk.values:

        print("\nMata kuliah tidak ditemukan.")

        print("\nDaftar mata kuliah yang tersedia:")

        for mk in df['mata_kuliah']:
            print("-", mk)

        return None


    # Mencari index mata kuliah
    index = daftar_mk[
        daftar_mk == nama_mk
    ].index[0]


    # Mengambil nilai similarity
    scores = list(
        enumerate(
            similarity[index]
        )
    )


    # Mengurutkan similarity
    # dari nilai terbesar ke terkecil
    scores = sorted(
        scores,
        key=lambda x: x[1],
        reverse=True
    )


    # Menyimpan hasil rekomendasi
    hasil = []


    # Mengambil jumlah rekomendasi
    for i, score in scores[1:jumlah + 1]:

        hasil.append({

            'Mata Kuliah':
                df.iloc[i]['mata_kuliah'],

            'Similarity':
                round(float(score), 3)

        })


    return pd.DataFrame(hasil)


# ==========================================================
# INPUT USER
# ==========================================================

print("\n================================================")
print("             INPUT MATA KULIAH")
print("================================================")

nama = input(
    "Masukkan nama mata kuliah: "
)


# ==========================================================
# HASIL REKOMENDASI
# ==========================================================

hasil = rekomendasi_mata_kuliah(
    nama,
    5
)


print("\n================================================")
print("             HASIL REKOMENDASI")
print("================================================")


if hasil is not None:

    print(
        hasil.to_string(index=False)
    )

    print("\nRekomendasi berhasil ditampilkan.")
```
